## Introduction
- Employee attrition is a critical challenge for organizations as it leads to loss of talent, increased hiring costs and reduced productivity.

- This project analyzes employee attrition data to identify key factors influencing why employees leave the company.

- The analysis is performed using Python, Pandas and Matplotlib.


## Dataset Overview
- The dataset contains employee-level information such as age, department, job satisfaction, overtime status, salary, promotion history and attrition status.

- It includes 1,470 employee records with 35 features and does not contain any missing values.

- Attrition is treated as the target variable where 1 represents employees who left the company and 0 represents employees who stayed.


## Problem Statement
- The goal of this project is to understand why employees leave the company and to identify how factors such as workload, job satisfaction and career growth affect employee retention.

```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("Hr.csv")
df[["Attrition", "Department", "MonthlyIncome", "JobSatisfaction"]].head()
```

```python
df.shape
df.info()
# Convert 'Attrition' column to numerical values (1 for 'Yes', 0 for 'No')
df['Attrition'] = df['Attrition'].map({'Yes': 1, 'No': 0})
df["Attrition"].value_counts(normalize=True)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("Hr.csv")
df['Attrition'] = df['Attrition'].map({'Yes': 1, 'No': 0})
df1=df.groupby("OverTime")["Attrition"].mean()
print(df1)
df2=df.groupby("Department")["Attrition"].mean()
print(df2)
df3=df.groupby("YearsSinceLastPromotion")["Attrition"].mean()
print(df3)
df4=df["YearsSinceLastPromotion"].value_counts().sort_index()
print(df4)
df5=df.groupby("JobSatisfaction")["Attrition"].mean()
print(df5)
df6=df.groupby(["OverTime", "JobSatisfaction"])["Attrition"].mean().unstack()
print(df6)
```

## Key Insights
- About 16 out of every 100 employees left the company so employee leaving is a real issue.

- Employees who work overtime leave much more often than those who do not.

- When employees are happier with their job they are less likely to leave.

- Employees who have not been promoted for many years tend to leave more.

- Employees who work overtime and are not very happy with their job leave the most.


```python
df["Attrition"].value_counts().plot(kind="bar")
plt.title("Employee Attrition Count")
plt.xlabel("Attrition (0 = Stayed, 1 = Left)")
plt.ylabel("Number of Employees")
plt.show()
```

```python
df.groupby("OverTime")["Attrition"].mean().plot(kind="bar")
plt.title("Attrition Rate by Overtime")
plt.ylabel("Attrition Rate")
plt.xlabel("OverTime")
plt.show()
```

```python
df.groupby("JobSatisfaction")["Attrition"].mean().plot(kind="bar")
plt.title("Attrition Rate by Job Satisfaction")
plt.xlabel("Job Satisfaction Level (1 = Low, 4 = Very High)")
plt.ylabel("Attrition Rate")
plt.show()
```

```python
pivot = df.groupby(["OverTime", "JobSatisfaction"])["Attrition"].mean().unstack()

pivot.plot(kind="bar", figsize=(8,5))
plt.title("Attrition Rate by Overtime and Job Satisfaction")
plt.ylabel("Attrition Rate")
plt.xlabel("OverTime")
plt.legend(title="Job Satisfaction")
plt.show()
```

## Business Recommendations
- Try to reduce extra working hours by sharing work better or hiring more people in busy teams.

- Keep an eye on how happy employees are especially those who work overtime so problems can be noticed early.

- Make sure employees get promotions on time and understand how they can grow in their careers.

- Pay more attention to departments where many employees are leaving especially the Sales team.


## Conclusion
This analysis shows that employees mainly leave because of too much work and low job happiness.
By reducing overtime, keeping employees happier and giving them timely promotions companies can reduce the number of people leaving.
